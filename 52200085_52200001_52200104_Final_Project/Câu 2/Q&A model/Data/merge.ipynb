```python
import json
from pathlib import Path

def merge_json_files(input_folder, output_file):
    merged_data = []

    # Duyệt tất cả các file JSON trong thư mục
    for json_file in Path(input_folder).glob('*.json'):
        try:
            with json_file.open('r', encoding='utf-8') as f:
                data = json.load(f)

                # Kiểm tra kiểu dữ liệu và xử lý tương ứng
                if isinstance(data, list):
                    merged_data.extend(data)
                elif isinstance(data, dict):
                    merged_data.append(data)
        except json.JSONDecodeError as e:
            print(f"Lỗi khi đọc file {json_file}: {e}")

    # Ghi dữ liệu hợp nhất vào file output
    with open(output_file, 'w', encoding='utf-8') as outfile:
        json.dump(merged_data, outfile, ensure_ascii=False, indent=4)

# Sử dụng hàm này
input_folder = 'Done/'
output_file = './merged_output.json'
merge_json_files(input_folder, output_file)
```

```python
def count_question_response_pairs(json_file):
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)

    pair_count = len([item for item in data if 'question' in item and 'response' in item])
    return pair_count

# Sử dụng hàm này
json_file = 'merged_output.json'
pair_count = count_question_response_pairs(json_file)
print(f"Tổng số cặp question và response là: {pair_count}")
```

```python
def find_duplicate_pairs(json_file):
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)

    seen = set()
    duplicates = []

    for item in data:
        if 'question' in item and 'response' in item:
            pair = (item['question'].strip(), item['response'].strip())
            if pair in seen:
                duplicates.append(item)
            else:
                seen.add(pair)

    return duplicates

duplicates = find_duplicate_pairs(json_file)
print(f"Số lượng cặp bị trùng lặp: {len(duplicates)}")
if duplicates:
    print("Ví dụ cặp trùng đầu tiên:")
    print(duplicates[0])
```
